```python
%py
import requests

# Source API URL
url = "https://public.tableau.com/app/sample-data/mobile_os_usage.csv"

# Perform API call
response = requests.get(url)

# Validate response
if response.status_code != 200:
    raise Exception(f"API call failed with status code {response.status_code}")

# Store response content in variable
api_data = response.text

print("API data successfully fetched.")
```

##### Write Data to Volume - usage_metrics


```python

# Target volume path
volume_path = "/Volumes/workspace/default/usage_metrics/mobile_os_usage.csv"

# Write data to volume
dbutils.fs.put(volume_path, api_data, overwrite=True)

print(f"Data successfully written to {volume_path}")
```
