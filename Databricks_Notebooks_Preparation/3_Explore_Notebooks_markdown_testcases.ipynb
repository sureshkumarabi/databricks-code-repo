############ Create PySpark DataFrame from Volume File


```python
%python
# File path in Databricks Volume
file_path = "/Volumes/workspace/default/usage_metrics/mobile_os_usage.csv"

# Read CSV into PySpark DataFrame
df1 = (
    spark.read
         .option("header", "true")
         .option("inferSchema", "true")
         .csv(file_path)
)

# Display sample data
df1.show(5)

# Print schema for verification
df1.printSchema()

```

######### Write DataFrame to Databricks Table

```python
# Rename columns to valid Delta table column names
df1_clean = (
    df1.withColumnRenamed("Mobile Operating System", "mobile_operating_system")
       .withColumnRenamed("Percent of Usage", "percent_of_usage")
)

# Write DataFrame to Databricks managed table
df1_clean.write.mode("overwrite").format("delta").saveAsTable("default.mobile_os_usage")

display(df1_clean)
```

```python
def to_upper_case(input_value):
    """
    Converts the given input string to upper case.

    :param input_value: str
    :return: str
    """
    if input_value is None:
        return None
    return str(input_value).upper()
```

```python
print(to_upper_case("hello"))
print(to_upper_case("databricks"))
```
